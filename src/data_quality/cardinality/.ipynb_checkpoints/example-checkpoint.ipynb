```python
import json

import pandas as pd
import polars as pl

from pyspark.sql import SparkSession
from pyspark.sql import types as T

from data_quality import (
    AdaptiveCardinalityConfig,
    identify_backend,
    profile_cardinality,
    profile_cardinality_json,
)

EXACT_CONFIG = AdaptiveCardinalityConfig(
    exact_unique_threshold=50_000,
    hll_precision=14,
)

FORCE_HLL_CONFIG = AdaptiveCardinalityConfig(
    exact_unique_threshold=100,
    hll_precision=14,
)
```

```python
#pandas
# ---------------------------------------------------------------------------
# pandas
# ---------------------------------------------------------------------------


def smoke_pandas_backend_detection() -> None:
    df = pd.DataFrame({"x": [1, 2, 3]})

    backend = identify_backend(df)

    assert backend.name.value == "pandas"
    print("PASS pandas backend detection")


def smoke_pandas_exact() -> None:
    df = pd.DataFrame(
        {
            "customer_id": pd.Series(
                [1, 2, 2, 3, None],
                dtype="Int64",
            )
        }
    )

    result = profile_cardinality(
        df,
        "customer_id",
        config=EXACT_CONFIG,
    )

    assert result.distinct_count == 3
    assert isinstance(result.distinct_count, int)
    assert result.is_exact is True
    assert result.final_mode.value == "exact"

    assert result.total_row_count == 5
    assert result.non_null_count == 4
    assert result.null_count == 1

    assert result.lineage["backend"]["name"] == "pandas"
    assert result.lineage["dtype_family"] == "integer"
    assert result.raw_hll_estimate is None

    print("PASS pandas exact")


def smoke_pandas_hll() -> None:
    expected = 10_000

    df = pd.DataFrame(
        {
            "customer_id": range(expected),
        }
    )

    result = profile_cardinality(
        df,
        "customer_id",
        config=FORCE_HLL_CONFIG,
    )

    assert result.is_exact is False
    assert result.final_mode.value == "hll"
    assert isinstance(result.distinct_count, int)

    assert result.promotion.occurred is True
    assert result.raw_hll_estimate is not None

    assert_hll_close(result.distinct_count, expected)

    print(
        "PASS pandas HLL",
        f"exact={expected}",
        f"estimate={result.distinct_count}",
    )


def smoke_pandas_dtype_kernels() -> None:
    df = pd.DataFrame(
        {
            "integer_col": pd.Series([1, 2, 3, 3], dtype="Int64"),
            "float_col": pd.Series([1.0, 2.0, -0.0, 0.0]),
            "bool_col": pd.Series([True, False, True], dtype="boolean"),
            "string_col": pd.Series(["a", "b", "a"], dtype="string"),
            "category_col": pd.Series(
                ["a", "b", "a"],
                dtype="category",
            ),
            "datetime_col": pd.to_datetime(
                ["2026-01-01", "2026-01-02", "2026-01-01"]
            ),
            "timedelta_col": pd.to_timedelta(
                ["1 day", "2 days", "1 day"]
            ),
            "period_col": pd.period_range(
                "2026-01",
                periods=4,
                freq="M",
            ),
            "interval_col": pd.arrays.IntervalArray.from_tuples(
                [(0, 1), (1, 2), (0, 1), (2, 3)]
            ),
        }
    )

    expected = {
        "integer_col": 3,
        "float_col": 3,       # -0.0 == 0.0
        "bool_col": 2,
        "string_col": 2,
        "category_col": 2,
        "datetime_col": 2,
        "timedelta_col": 2,
        "period_col": 4,
        "interval_col": 3,
    }

    for column, expected_count in expected.items():
        result = profile_cardinality(df, column)

        assert result.distinct_count == expected_count
        assert result.is_exact is True
        assert result.lineage["scalar_fallback_used"] is False

    print("PASS pandas dtype-specialized kernels")


def smoke_pandas_object_fallback() -> None:
    df = pd.DataFrame(
        {
            "mixed": pd.Series(
                [1, True, 1.0, "1"],
                dtype="object",
            )
        }
    )

    result = profile_cardinality(df, "mixed")

    # Project canonicalization intentionally distinguishes these types.
    assert result.distinct_count == 4
    assert result.lineage["scalar_fallback_used"] is True

    warning_codes = [warning.code.value for warning in result.warnings]
    assert "scalar_canonicalization_fallback" in warning_codes

    print("PASS pandas object fallback")


def smoke_pandas_unsupported_column_isolation() -> None:
    df = pd.DataFrame(
        {
            "id": [1, 2, 3],
            "name": ["a", "b", "a"],
            "unsupported": [[1], [2], [3]],
        }
    )

    report = profile_cardinality(df)

    assert report.total_row_count == 3
    assert report.column_count == 3
    assert report.successful_column_count == 2
    assert report.unsupported_column_count == 1

    outcomes = {outcome.column: outcome for outcome in report.columns}

    assert outcomes["id"].status == "ok"
    assert outcomes["name"].status == "ok"
    assert outcomes["unsupported"].status == "unsupported"
    assert outcomes["unsupported"].error is not None

    print("PASS pandas unsupported-column isolation")


def smoke_pandas_json() -> None:
    df = pd.DataFrame(
        {
            "customer_id": [1, 2, 2, 3],
        }
    )

    payload = profile_cardinality_json(
        df,
        "customer_id",
    )

    parsed = json.loads(payload)

    assert parsed["output"]["distinct_count"] == 3
    assert parsed["output"]["is_exact"] is True
    assert parsed["lineage"]["backend"]["name"] == "pandas"
    assert "parameters" in parsed
    assert "promotion" in parsed
    assert "hll_state" in parsed

    print("PASS pandas JSON lineage")


def run_pandas_smokes() -> None:
    print("\n=== PANDAS ===")

    smoke_pandas_backend_detection()
    smoke_pandas_exact()
    smoke_pandas_hll()
    smoke_pandas_dtype_kernels()
    smoke_pandas_object_fallback()
    smoke_pandas_unsupported_column_isolation()
    smoke_pandas_json()
```

```python
# ---------------------------------------------------------------------------
# Polars
# ---------------------------------------------------------------------------


def smoke_polars_backend_detection() -> None:
    df = pl.DataFrame({"x": [1, 2, 3]})

    backend = identify_backend(df)

    assert backend.name.value == "polars"
    print("PASS Polars backend detection")


def smoke_polars_exact() -> None:
    df = pl.DataFrame(
        {
            "customer_id": [1, 2, 2, 3, None],
        }
    )

    result = profile_cardinality(
        df,
        "customer_id",
        config=EXACT_CONFIG,
    )

    assert result.distinct_count == 3
    assert isinstance(result.distinct_count, int)
    assert result.is_exact is True

    assert result.total_row_count == 5
    assert result.non_null_count == 4
    assert result.null_count == 1

    assert result.lineage["backend"]["name"] == "polars"
    assert result.lineage["dtype_family"] == "integer"

    print("PASS Polars exact")


def smoke_polars_hll() -> None:
    expected = 10_000

    df = pl.DataFrame(
        {
            "customer_id": range(expected),
        }
    )

    result = profile_cardinality(
        df,
        "customer_id",
        config=FORCE_HLL_CONFIG,
    )

    assert result.is_exact is False
    assert result.final_mode.value == "hll"
    assert isinstance(result.distinct_count, int)

    assert result.promotion.occurred is True
    assert result.raw_hll_estimate is not None

    assert_hll_close(result.distinct_count, expected)

    print(
        "PASS Polars HLL",
        f"exact={expected}",
        f"estimate={result.distinct_count}",
    )


def smoke_polars_lazy_exact() -> None:
    df = pl.DataFrame(
        {
            "customer_id": [1, 2, 2, 3, None],
        }
    ).lazy()

    result = profile_cardinality(
        df,
        "customer_id",
    )

    assert result.distinct_count == 3
    assert result.is_exact is True
    assert result.lineage["backend"]["name"] == "polars"

    print("PASS Polars LazyFrame exact")


def smoke_polars_lazy_hll() -> None:
    expected = 10_000

    df = pl.DataFrame(
        {
            "customer_id": range(expected),
        }
    ).lazy()

    result = profile_cardinality(
        df,
        "customer_id",
        config=FORCE_HLL_CONFIG,
    )

    assert result.is_exact is False
    assert result.promotion.occurred is True

    assert_hll_close(result.distinct_count, expected)

    print("PASS Polars LazyFrame HLL")


def smoke_polars_dtypes() -> None:
    df = pl.DataFrame(
        {
            "integer_col": [1, 2, 3, 3],
            "float_col": [1.0, 2.0, -0.0, 0.0],
            "bool_col": [True, False, True, False],
            "string_col": ["a", "b", "a", "c"],
            "binary_col": [b"a", b"b", b"a", b"c"],
            "date_col": [
                pd.Timestamp("2026-01-01").date(),
                pd.Timestamp("2026-01-02").date(),
                pd.Timestamp("2026-01-01").date(),
                pd.Timestamp("2026-01-03").date(),
            ],
        }
    )

    expected = {
        "integer_col": 3,
        "float_col": 3,
        "bool_col": 2,
        "string_col": 3,
        "binary_col": 3,
        "date_col": 3,
    }

    for column, expected_count in expected.items():
        result = profile_cardinality(df, column)

        assert result.distinct_count == expected_count
        assert result.is_exact is True
        assert result.lineage["scalar_fallback_used"] is False

    print("PASS Polars dtype kernels")


def smoke_polars_nan_and_null() -> None:
    df = pl.DataFrame(
        {
            "x": [1.0, float("nan"), None, 2.0, -0.0, 0.0],
        }
    )

    result = profile_cardinality(df, "x")

    # NaN and None excluded, signed zeros normalized together.
    assert result.distinct_count == 3
    assert result.null_count == 2

    print("PASS Polars NaN/null/signed-zero semantics")


def smoke_polars_unsupported_nested() -> None:
    df = pl.DataFrame(
        {
            "id": [1, 2, 3],
            "nested": [[1, 2], [3], [4, 5]],
        }
    )

    report = profile_cardinality(df)

    outcomes = {outcome.column: outcome for outcome in report.columns}

    assert outcomes["id"].status == "ok"
    assert outcomes["nested"].status == "unsupported"

    assert report.total_row_count == 3
    assert report.successful_column_count == 1
    assert report.unsupported_column_count == 1

    print("PASS Polars nested-column isolation")


def smoke_polars_all_unsupported_row_count() -> None:
    df = pl.DataFrame(
        {
            "a": [[1], [2], [3]],
            "b": [{"x": 1}, {"x": 2}, {"x": 3}],
        }
    )

    report = profile_cardinality(df)

    assert report.total_row_count == 3
    assert report.successful_column_count == 0
    assert report.unsupported_column_count == 2

    print("PASS Polars all-unsupported row-count handling")


def smoke_polars_json() -> None:
    df = pl.DataFrame(
        {
            "customer_id": [1, 2, 2, 3],
        }
    )

    parsed = json.loads(
        profile_cardinality_json(
            df,
            "customer_id",
        )
    )

    assert parsed["output"]["distinct_count"] == 3
    assert parsed["lineage"]["backend"]["name"] == "polars"

    print("PASS Polars JSON lineage")


def run_polars_smokes() -> None:
    print("\n=== POLARS ===")

    smoke_polars_backend_detection()
    smoke_polars_exact()
    smoke_polars_hll()
    smoke_polars_lazy_exact()
    smoke_polars_lazy_hll()
    smoke_polars_dtypes()
    smoke_polars_nan_and_null()
    smoke_polars_unsupported_nested()
    smoke_polars_all_unsupported_row_count()
    smoke_polars_json()
```

```python
# ---------------------------------------------------------------------------
# PySpark
# ---------------------------------------------------------------------------


def smoke_pyspark_backend_detection(spark: SparkSession) -> None:
    df = spark.createDataFrame([(1,), (2,), (3,)], ["x"])

    backend = identify_backend(df)

    assert backend.name.value == "pyspark"
    print("PASS PySpark backend detection")


def smoke_pyspark_exact(spark: SparkSession) -> None:
    df = spark.createDataFrame(
        [
            (1,),
            (2,),
            (2,),
            (3,),
            (None,),
        ],
        schema=T.StructType(
            [
                T.StructField(
                    "customer_id",
                    T.LongType(),
                    nullable=True,
                )
            ]
        ),
    )

    result = profile_cardinality(
        df,
        "customer_id",
        config=EXACT_CONFIG,
    )

    assert result.distinct_count == 3
    assert isinstance(result.distinct_count, int)
    assert result.is_exact is True

    assert result.total_row_count == 5
    assert result.non_null_count == 4
    assert result.null_count == 1

    assert result.lineage["backend"]["name"] == "pyspark"
    assert result.lineage["dtype_family"] == "integer"
    assert result.lineage["distributed_execution"] is True

    print("PASS PySpark exact")


def smoke_pyspark_hll(spark: SparkSession) -> None:
    expected = 10_000

    df = spark.range(expected).withColumnRenamed(
        "id",
        "customer_id",
    )

    result = profile_cardinality(
        df,
        "customer_id",
        config=FORCE_HLL_CONFIG,
    )

    assert result.is_exact is False
    assert result.final_mode.value == "hll"
    assert isinstance(result.distinct_count, int)

    assert result.promotion.occurred is True
    assert result.promotion.row_position is None
    assert result.raw_hll_estimate is not None

    assert result.lineage["distributed_execution"] is True
    assert result.lineage["adaptive_strategy"] == (
        "two_stage_exact_then_hll"
    )

    assert_hll_close(result.distinct_count, expected)

    print(
        "PASS PySpark distributed HLL",
        f"exact={expected}",
        f"estimate={result.distinct_count}",
    )


def smoke_pyspark_dtypes(spark: SparkSession) -> None:
    schema = T.StructType(
        [
            T.StructField("integer_col", T.LongType(), True),
            T.StructField("float_col", T.DoubleType(), True),
            T.StructField("bool_col", T.BooleanType(), True),
            T.StructField("string_col", T.StringType(), True),
            T.StructField("binary_col", T.BinaryType(), True),
        ]
    )

    df = spark.createDataFrame(
        [
            (1, 1.0, True, "a", b"a"),
            (2, 2.0, False, "b", b"b"),
            (3, -0.0, True, "a", b"a"),
            (3, 0.0, False, "c", b"c"),
        ],
        schema=schema,
    )

    expected = {
        "integer_col": 3,
        "float_col": 3,
        "bool_col": 2,
        "string_col": 3,
        "binary_col": 3,
    }

    for column, expected_count in expected.items():
        result = profile_cardinality(df, column)

        assert result.distinct_count == expected_count
        assert result.is_exact is True

    print("PASS PySpark dtype families")


def smoke_pyspark_nan_null_zero(spark: SparkSession) -> None:
    schema = T.StructType(
        [
            T.StructField(
                "x",
                T.DoubleType(),
                True,
            )
        ]
    )

    df = spark.createDataFrame(
        [
            (1.0,),
            (float("nan"),),
            (None,),
            (2.0,),
            (-0.0,),
            (0.0,),
        ],
        schema=schema,
    )

    result = profile_cardinality(df, "x")

    # NaN and null excluded; -0.0 / 0.0 normalize together.
    assert result.distinct_count == 3
    assert result.null_count == 2

    print("PASS PySpark NaN/null/signed-zero semantics")


def smoke_pyspark_unsupported_nested(spark: SparkSession) -> None:
    schema = T.StructType(
        [
            T.StructField("id", T.LongType(), False),
            T.StructField(
                "nested",
                T.ArrayType(T.LongType()),
                True,
            ),
        ]
    )

    df = spark.createDataFrame(
        [
            (1, [1, 2]),
            (2, [3]),
            (3, [4, 5]),
        ],
        schema=schema,
    )

    report = profile_cardinality(df)

    outcomes = {
        outcome.column: outcome
        for outcome in report.columns
    }

    assert outcomes["id"].status == "ok"
    assert outcomes["nested"].status == "unsupported"

    assert report.total_row_count == 3
    assert report.successful_column_count == 1
    assert report.unsupported_column_count == 1

    print("PASS PySpark nested-column isolation")


def smoke_pyspark_all_unsupported(spark: SparkSession) -> None:
    schema = T.StructType(
        [
            T.StructField(
                "array_col",
                T.ArrayType(T.IntegerType()),
                True,
            ),
            T.StructField(
                "map_col",
                T.MapType(
                    T.StringType(),
                    T.IntegerType(),
                ),
                True,
            ),
        ]
    )

    df = spark.createDataFrame(
        [
            ([1], {"a": 1}),
            ([2], {"b": 2}),
            ([3], {"c": 3}),
        ],
        schema=schema,
    )

    report = profile_cardinality(df)

    assert report.total_row_count == 3
    assert report.successful_column_count == 0
    assert report.unsupported_column_count == 2

    print("PASS PySpark all-unsupported row count")


def smoke_pyspark_json(spark: SparkSession) -> None:
    df = spark.createDataFrame(
        [
            (1,),
            (2,),
            (2,),
            (3,),
        ],
        ["customer_id"],
    )

    parsed = json.loads(
        profile_cardinality_json(
            df,
            "customer_id",
        )
    )

    assert parsed["output"]["distinct_count"] == 3
    assert parsed["lineage"]["backend"]["name"] == "pyspark"
    assert parsed["lineage"]["distributed_execution"] is True

    print("PASS PySpark JSON lineage")


def run_pyspark_smokes() -> None:
    print("\n=== PYSPARK ===")

    spark = (
        SparkSession.builder
        .master("local[2]")
        .appName("data-quality-cardinality-smoke")
        .config("spark.ui.enabled", "false")
        .getOrCreate()
    )

    spark.sparkContext.setLogLevel("ERROR")

    try:
        smoke_pyspark_backend_detection(spark)
        smoke_pyspark_exact(spark)
        smoke_pyspark_hll(spark)
        smoke_pyspark_dtypes(spark)
        smoke_pyspark_nan_null_zero(spark)
        smoke_pyspark_unsupported_nested(spark)
        smoke_pyspark_all_unsupported(spark)
        smoke_pyspark_json(spark)
    finally:
        spark.stop()
```

```python
# ---------------------------------------------------------------------------
# Cross-platform consistency
# ---------------------------------------------------------------------------


def smoke_cross_platform_exact_consistency() -> None:
    values = [1, 2, 2, 3, 4, 4, 5]

    pandas_df = pd.DataFrame({"id": values})
    polars_df = pl.DataFrame({"id": values})

    pandas_result = profile_cardinality(
        pandas_df,
        "id",
    )

    polars_result = profile_cardinality(
        polars_df,
        "id",
    )

    assert pandas_result.distinct_count == 5
    assert polars_result.distinct_count == 5

    print("PASS pandas/Polars exact consistency")


def main() -> None:
    run_pandas_smokes()
    run_polars_smokes()
    run_pyspark_smokes()
    smoke_cross_platform_exact_consistency()

    print("\n================================")
    print("ALL PLATFORM SMOKE TESTS PASSED")
    print("================================")
```
