```python
import inspect
import os
import time
from decimal import Decimal, localcontext

import numpy as np
import pandas as pd

import data_quality
import data_quality.cardinality.hashing as hashing
import data_quality.cardinality.adaptive as adaptive
import data_quality.cardinality.pandas_backend as pandas_backend

from data_quality import AdaptiveCardinalityConfig, profile_cardinality
from data_quality.cardinality.hashing import canonicalize_scalar
from data_quality.cardinality.pandas_backend import (
    iter_pandas_hash_arrays,
    pandas_hll_sketch,
)

print("data_quality:", data_quality.__file__)
print("hashing:", hashing.__file__)
print("adaptive:", adaptive.__file__)
print("pandas_backend:", pandas_backend.__file__)
print("cwd:", os.getcwd())
```

# Decimal canonicalisation

## 2. Equivalent Decimal values

```python
a = canonicalize_scalar(Decimal("1.2300"))
b = canonicalize_scalar(Decimal("1.23"))

print(a)
print(b)

assert a == b
assert a == b"d0:123:-2"

print("PASS: Decimal trailing-zero canonicalisation")
```

## 3. Decimal context independence

```python
value = Decimal("123456789.123450000")
expected = canonicalize_scalar(value)

for precision in [2, 5, 10, 28, 50]:
    with localcontext() as ctx:
        ctx.prec = precision
        actual = canonicalize_scalar(value)

    print("precision:", precision, "same:", actual == expected)
    assert actual == expected

print("PASS: Decimal canonicalisation is context-independent")
```

## 4. Huge Decimal coefficient

```python
huge = Decimal("123" + ("0" * 10_000) + "E-10002")
small = Decimal("1.23")

huge_token = canonicalize_scalar(huge)
small_token = canonicalize_scalar(small)

print("Huge token length:", len(huge_token))
print("Canonical token:", huge_token)

assert huge_token == small_token
assert small_token == b"d0:123:-2"

print("PASS: huge Decimal canonicalisation")
```

# Signed / unsigned integer semantics

## 5. Full UInt64 domain

```python
values = [
    0,
    2**63 - 1,
    2**63,
    2**64 - 1,
]

df = pd.DataFrame({
    "x": pd.Series(values, dtype="UInt64")
})

result = profile_cardinality(df, "x")

print("Count:", result.distinct_count)
print("Exact:", result.is_exact)
print("dtype family:", result.lineage.get("dtype_family"))
print("kernel:", result.lineage.get("kernel"))

assert result.distinct_count == 4
assert result.is_exact is True

print("PASS: full UInt64 domain preserved")
```

## 6. Signed and unsigned sketch contracts differ

```python
signed_df = pd.DataFrame({
    "x": pd.Series([1, 2, 3], dtype="int64")
})

unsigned_df = pd.DataFrame({
    "x": pd.Series([1, 2, 3], dtype="uint64")
})

signed_sketch = pandas_hll_sketch(signed_df, "x")
unsigned_sketch = pandas_hll_sketch(unsigned_df, "x")

print("Signed:", signed_sketch.hash_configuration)
print("Unsigned:", unsigned_sketch.hash_configuration)

assert signed_sketch.hash_configuration != unsigned_sketch.hash_configuration

try:
    signed_sketch.merge(unsigned_sketch)
except ValueError as exc:
    print("Expected merge rejection:")
    print(exc)
else:
    raise AssertionError("Signed/unsigned sketches should not merge")

print("PASS: signed/unsigned contracts remain distinct")
```

# Temporal normalization

## 7. Datetime units normalize to identical hashes

```python
logical_values = np.array([
    "2026-01-01T00:00:01",
    "2026-01-01T00:00:02",
    "2026-01-01T00:00:03",
])

def datetime_hashes(unit):
    arr = logical_values.astype(f"datetime64[{unit}]")
    frame = pd.DataFrame({"x": pd.Series(arr)})
    return np.concatenate(list(iter_pandas_hash_arrays(frame, "x")))

hash_s = datetime_hashes("s")
hash_ms = datetime_hashes("ms")
hash_us = datetime_hashes("us")
hash_ns = datetime_hashes("ns")

print("s :", hash_s)
print("ms:", hash_ms)
print("us:", hash_us)
print("ns:", hash_ns)

np.testing.assert_array_equal(hash_s, hash_ns)
np.testing.assert_array_equal(hash_ms, hash_ns)
np.testing.assert_array_equal(hash_us, hash_ns)

print("PASS: datetime units normalize to nanoseconds")
```

## 8. Timedelta units normalize to identical hashes

```python
def td_frame(values, unit):
    return pd.DataFrame({
        "x": pd.Series(
            np.array(values, dtype=f"timedelta64[{unit}]")
        )
    })

frames = {
    "s": td_frame([1, 2, 3], "s"),
    "ms": td_frame([1000, 2000, 3000], "ms"),
    "us": td_frame([1_000_000, 2_000_000, 3_000_000], "us"),
    "ns": td_frame(
        [1_000_000_000, 2_000_000_000, 3_000_000_000],
        "ns",
    ),
}

hashes = {
    unit: np.concatenate(list(iter_pandas_hash_arrays(frame, "x")))
    for unit, frame in frames.items()
}

for unit, arr in hashes.items():
    print(unit, arr)

np.testing.assert_array_equal(hashes["s"], hashes["ns"])
np.testing.assert_array_equal(hashes["ms"], hashes["ns"])
np.testing.assert_array_equal(hashes["us"], hashes["ns"])

print("PASS: timedelta units normalize to nanoseconds")
```

# Categorical identity

## 9. Category dictionary order is irrelevant

```python
force_hll = AdaptiveCardinalityConfig(
    exact_unique_threshold=2,
    hll_precision=14,
)

df_a = pd.DataFrame({
    "x": pd.Categorical(
        ["a", "b", "c", "a", "b", "c"],
        categories=["a", "b", "c"],
    )
})

df_b = pd.DataFrame({
    "x": pd.Categorical(
        ["a", "b", "c", "a", "b", "c"],
        categories=["c", "b", "a"],
    )
})

a = profile_cardinality(df_a, "x", config=force_hll)
b = profile_cardinality(df_b, "x", config=force_hll)

print("A:", a.distinct_count, a.raw_hll_estimate)
print("B:", b.distinct_count, b.raw_hll_estimate)
print("A kernel:", a.lineage.get("kernel"))
print("B kernel:", b.lineage.get("kernel"))

assert a.distinct_count == 3
assert b.distinct_count == 3
assert a.raw_hll_estimate == b.raw_hll_estimate

print("PASS: category dictionary order is irrelevant")
```

## 10. ordered=True / False is irrelevant to cardinality identity

```python
df_unordered = pd.DataFrame({
    "x": pd.Categorical(
        ["low", "medium", "high", "low"],
        categories=["low", "medium", "high"],
        ordered=False,
    )
})

df_ordered = pd.DataFrame({
    "x": pd.Categorical(
        ["low", "medium", "high", "low"],
        categories=["high", "medium", "low"],
        ordered=True,
    )
})

u = profile_cardinality(df_unordered, "x", config=force_hll)
o = profile_cardinality(df_ordered, "x", config=force_hll)

print("unordered:", u.distinct_count, u.raw_hll_estimate)
print("ordered:", o.distinct_count, o.raw_hll_estimate)

assert u.distinct_count == 3
assert o.distinct_count == 3
assert u.raw_hll_estimate == o.raw_hll_estimate

print("PASS: ordered flag does not affect cardinality identity")
```

## 11. Different category dictionaries remain merge-compatible

```python
df_a = pd.DataFrame({
    "x": pd.Categorical(
        ["a", "b", "a", "b"],
        categories=["a", "b"],
    )
})

df_b = pd.DataFrame({
    "x": pd.Categorical(
        ["b", "c", "b", "c"],
        categories=["b", "c"],
    )
})

sketch_a = pandas_hll_sketch(df_a, "x")
sketch_b = pandas_hll_sketch(df_b, "x")

print("A:", sketch_a.hash_configuration)
print("B:", sketch_b.hash_configuration)

assert sketch_a.hash_configuration == sketch_b.hash_configuration

merged = sketch_a.copy()
merged.merge(sketch_b)

print("Merged estimate:", merged.estimate())
print("PASS: categorical sketches portable across dictionaries")
```

## 12. Large categorical strings

```python
huge_a = "A" * 1_000_000
huge_b = "B" * 1_000_000

df = pd.DataFrame({
    "x": pd.Categorical([
        huge_a,
        huge_b,
        huge_a,
        huge_b,
        huge_a,
    ])
})

start = time.perf_counter()
result = profile_cardinality(df, "x")
elapsed = time.perf_counter() - start

print("Distinct:", result.distinct_count)
print("Exact:", result.is_exact)
print(f"Elapsed: {elapsed:.4f}s")

assert result.distinct_count == 2
assert result.is_exact is True

print("PASS: large categorical values")
```

# Adaptive exact-state performance / correctness

## 15. Duplicate-heavy workload stays exact

```python
values = np.tile(
    np.arange(1_000, dtype=np.int64),
    2_000,
)

df = pd.DataFrame({"x": values})

config = AdaptiveCardinalityConfig(
    exact_unique_threshold=50_000,
    hll_precision=14,
)

start = time.perf_counter()
result = profile_cardinality(df, "x", config=config)
elapsed = time.perf_counter() - start

print("Rows:", len(df))
print("Expected unique:", df["x"].nunique())
print("Result:", result.distinct_count)
print("Exact:", result.is_exact)
print("Promotion:", result.promotion)
print(f"Elapsed: {elapsed:.4f}s")

assert result.distinct_count == 1_000
assert result.is_exact is True
assert result.promotion.occurred is False

print("PASS: duplicate-heavy exact fast path")
```

## 16. New values still trigger exact → HLL promotion

```python
duplicates = np.tile(
    np.arange(500, dtype=np.int64),
    100,
)

new_values = np.arange(
    500,
    10_000,
    dtype=np.int64,
)

values = np.concatenate([duplicates, new_values])
df = pd.DataFrame({"x": values})

config = AdaptiveCardinalityConfig(
    exact_unique_threshold=1_000,
    hll_precision=14,
)

result = profile_cardinality(df, "x", config=config)

expected = int(df["x"].nunique())
relative_error = abs(result.distinct_count - expected) / expected

print("True unique:", expected)
print("Estimate:", result.distinct_count)
print("Exact:", result.is_exact)
print("Promotion:", result.promotion)
print("Raw estimate:", result.raw_hll_estimate)
print(f"Relative error: {relative_error:.4%}")

assert result.is_exact is False
assert result.promotion.occurred is True
assert result.raw_hll_estimate is not None
assert relative_error < 0.05

print("PASS: new values still trigger promotion")
```

## 17. Duplicate-heavy prefix followed by threshold-crossing values

```python
prefix = np.tile(
    np.arange(900, dtype=np.int64),
    30,
)

tail = np.concatenate([
    np.arange(0, 900, dtype=np.int64),
    np.arange(900, 1_100, dtype=np.int64),
])

df = pd.DataFrame({
    "x": np.concatenate([prefix, tail])
})

config = AdaptiveCardinalityConfig(
    exact_unique_threshold=1_000,
    threshold_check_interval=256,
    hll_precision=14,
)

result = profile_cardinality(df, "x", config=config)

print("True unique:", df["x"].nunique())
print("Estimate:", result.distinct_count)
print("Promotion:", result.promotion)

assert result.is_exact is False
assert result.promotion.occurred is True

print("PASS: mixed duplicate/new-value promotion boundary")
```
