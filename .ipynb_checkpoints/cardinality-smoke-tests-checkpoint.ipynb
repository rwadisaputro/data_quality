# Data Quality Cardinality Platform Smoke Tests — Run-All Safe

This notebook tests pandas, Polars, and PySpark.

The PySpark section is **disabled by default** so `Run All` cannot hang indefinitely during JVM startup.
When you enable it, Spark first starts in a separate subprocess with a **60-second timeout**. Only if that succeeds does the notebook start a real Spark session.


## Setup

```python
# If needed, uncomment, run, then restart the kernel:
# %pip install -e ".[all-backends,test]"
```

```python
import json
import os
import shutil
import subprocess
import sys
import importlib.util

import pandas as pd
import polars as pl

from data_quality import (
    AdaptiveCardinalityConfig,
    identify_backend,
    profile_cardinality,
    profile_cardinality_json,
)

EXACT_CONFIG = AdaptiveCardinalityConfig(
    exact_unique_threshold=50_000,
    hll_precision=14,
)

FORCE_HLL_CONFIG = AdaptiveCardinalityConfig(
    exact_unique_threshold=100,
    hll_precision=14,
)

def assert_hll_close(actual, expected, tolerance=0.05):
    error = abs(actual - expected) / expected
    print("Expected:", expected)
    print("Estimated:", actual)
    print(f"Relative error: {error:.4%}")
    assert error < tolerance

print("Python:", sys.version.split()[0])
print("pandas:", pd.__version__)
print("polars:", pl.__version__)
```

# pandas

```python
df = pd.DataFrame({"x": [1, 2, 3]})
backend = identify_backend(df)
print(backend)
assert backend.name.value == "pandas"
print("PASS: pandas backend detection")
```

```python
df = pd.DataFrame({
    "customer_id": pd.Series([1, 2, 2, 3, None], dtype="Int64")
})

result = profile_cardinality(df, "customer_id", config=EXACT_CONFIG)

print("Count:", result.distinct_count)
print("Exact:", result.is_exact)
print("Rows:", result.total_row_count)
print("Null:", result.null_count)

assert result.distinct_count == 3
assert result.is_exact is True
assert result.total_row_count == 5
assert result.null_count == 1

print("PASS: pandas exact")
```

```python
expected = 10_000
df = pd.DataFrame({"customer_id": range(expected)})

result = profile_cardinality(
    df,
    "customer_id",
    config=FORCE_HLL_CONFIG,
)

print("Estimate:", result.distinct_count)
print("Promotion:", result.promotion)

assert result.is_exact is False
assert result.promotion.occurred is True
assert_hll_close(result.distinct_count, expected)

print("PASS: pandas HLL")
```

```python
df = pd.DataFrame({
    "id": [1, 2, 3],
    "unsupported": [[1], [2], [3]],
})

report = profile_cardinality(df)

print("Rows:", report.total_row_count)
print("Successful:", report.successful_column_count)
print("Unsupported:", report.unsupported_column_count)

assert report.total_row_count == 3
assert report.successful_column_count == 1
assert report.unsupported_column_count == 1

print("PASS: pandas unsupported-column isolation")
```

# PySpark — safe gated section

`RUN_PYSPARK` is `False` by default.

For normal `Run All`, Spark is skipped.

To test Spark:
1. Set `RUN_PYSPARK = True`.
2. Run the PySpark cells from this point onward.
3. The isolated startup check times out after 60 seconds instead of hanging Jupyter.


```python
RUN_PYSPARK = False

print("RUN_PYSPARK =", RUN_PYSPARK)

PYSPARK_AVAILABLE = False
SPARK_PREFLIGHT_OK = False
spark = None
```

## PySpark diagnostics

```python
if RUN_PYSPARK:
    spec = importlib.util.find_spec("pyspark")
    java_path = shutil.which("java")

    print("Python executable:", sys.executable)
    print("JAVA_HOME:", os.environ.get("JAVA_HOME"))
    print("java executable:", java_path)
    print("pyspark origin:", spec.origin if spec else None)
    print(
        "pyspark package locations:",
        spec.submodule_search_locations if spec else None,
    )

    assert spec is not None, "PySpark is not installed."
    assert spec.submodule_search_locations is not None, (
        "Imported a plain pyspark module instead of the real package."
    )
    assert java_path is not None, "Java is not available on PATH."

    java = subprocess.run(
        ["java", "-version"],
        capture_output=True,
        text=True,
        timeout=10,
    )

    print(java.stdout)
    print(java.stderr)

    PYSPARK_AVAILABLE = True
else:
    print("SKIP: PySpark diagnostics")
```

## Isolated Spark startup preflight — 60 second hard timeout

```python
if RUN_PYSPARK and PYSPARK_AVAILABLE:
    preflight_code = r'''
import os
os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"

from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .master("local[1]")
    .appName("data-quality-preflight")
    .config("spark.ui.enabled", "false")
    .config("spark.driver.host", "127.0.0.1")
    .config("spark.driver.bindAddress", "127.0.0.1")
    .getOrCreate()
)

assert spark.range(1).count() == 1
print("SPARK_PREFLIGHT_OK")
spark.stop()
'''

    env = os.environ.copy()
    env["SPARK_LOCAL_IP"] = "127.0.0.1"

    print("Starting Spark preflight...")

    try:
        proc = subprocess.run(
            [sys.executable, "-c", preflight_code],
            capture_output=True,
            text=True,
            timeout=60,
            env=env,
        )

        print("Return code:", proc.returncode)

        if proc.stdout:
            print("stdout:")
            print(proc.stdout)

        if proc.stderr:
            print("stderr:")
            print(proc.stderr)

        SPARK_PREFLIGHT_OK = (
            proc.returncode == 0
            and "SPARK_PREFLIGHT_OK" in proc.stdout
        )

    except subprocess.TimeoutExpired:
        SPARK_PREFLIGHT_OK = False
        print("FAIL: Spark preflight exceeded 60 seconds and was killed.")

    print("SPARK_PREFLIGHT_OK =", SPARK_PREFLIGHT_OK)
else:
    print("SKIP: Spark startup preflight")
```

## Start Spark only after preflight passes

```python
if RUN_PYSPARK and SPARK_PREFLIGHT_OK:
    os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"

    from pyspark.sql import SparkSession
    from pyspark.sql import types as T

    spark = (
        SparkSession.builder
        .master("local[2]")
        .appName("data-quality-cardinality-smoke")
        .config("spark.ui.enabled", "false")
        .config("spark.driver.host", "127.0.0.1")
        .config("spark.driver.bindAddress", "127.0.0.1")
        .getOrCreate()
    )

    spark.sparkContext.setLogLevel("ERROR")

    print("Spark version:", spark.version)
    print("Spark session started")
else:
    print("SKIP: Spark session start")
```

## PySpark backend detection

```python
if spark is not None:
    df = spark.createDataFrame([(1,), (2,), (3,)], ["x"])

    backend = identify_backend(df)
    print(backend)

    assert backend.name.value == "pyspark"
    print("PASS: PySpark backend detection")
else:
    print("SKIP: PySpark backend detection")
```

## PySpark exact cardinality

```python
if spark is not None:
    schema = T.StructType([
        T.StructField("customer_id", T.LongType(), True)
    ])

    df = spark.createDataFrame(
        [(1,), (2,), (2,), (3,), (None,)],
        schema=schema,
    )

    result = profile_cardinality(
        df,
        "customer_id",
        config=EXACT_CONFIG,
    )

    print("Count:", result.distinct_count)
    print("Exact:", result.is_exact)
    print("Rows:", result.total_row_count)
    print("Null:", result.null_count)

    assert result.distinct_count == 3
    assert result.is_exact is True
    assert result.total_row_count == 5
    assert result.null_count == 1

    print("PASS: PySpark exact")
else:
    print("SKIP: PySpark exact")
```

## PySpark distributed HLL promotion

```python
if spark is not None:
    expected = 10_000

    df = spark.range(expected).withColumnRenamed("id", "customer_id")

    result = profile_cardinality(
        df,
        "customer_id",
        config=FORCE_HLL_CONFIG,
    )

    print("Estimate:", result.distinct_count)
    print("Promotion:", result.promotion)
    print("Raw estimate:", result.raw_hll_estimate)

    assert result.is_exact is False
    assert result.promotion.occurred is True
    assert result.raw_hll_estimate is not None

    assert_hll_close(result.distinct_count, expected)

    print("PASS: PySpark HLL")
else:
    print("SKIP: PySpark HLL")
```

## PySpark NaN / null / signed-zero semantics

```python
if spark is not None:
    schema = T.StructType([
        T.StructField("x", T.DoubleType(), True)
    ])

    df = spark.createDataFrame(
        [
            (1.0,),
            (float("nan"),),
            (None,),
            (2.0,),
            (-0.0,),
            (0.0,),
        ],
        schema=schema,
    )

    result = profile_cardinality(df, "x")

    print("Count:", result.distinct_count)
    print("Null-like:", result.null_count)

    assert result.distinct_count == 3
    assert result.null_count == 2

    print("PASS: PySpark NaN/null/signed-zero")
else:
    print("SKIP: PySpark NaN/null/signed-zero")
```

## PySpark scalar dtype families

```python
if spark is not None:
    schema = T.StructType([
        T.StructField("integer_col", T.LongType(), True),
        T.StructField("float_col", T.DoubleType(), True),
        T.StructField("bool_col", T.BooleanType(), True),
        T.StructField("string_col", T.StringType(), True),
        T.StructField("binary_col", T.BinaryType(), True),
    ])

    df = spark.createDataFrame(
        [
            (1, 1.0, True, "a", b"a"),
            (2, 2.0, False, "b", b"b"),
            (3, -0.0, True, "a", b"a"),
            (3, 0.0, False, "c", b"c"),
        ],
        schema=schema,
    )

    expected = {
        "integer_col": 3,
        "float_col": 3,
        "bool_col": 2,
        "string_col": 3,
        "binary_col": 3,
    }

    for column, expected_count in expected.items():
        result = profile_cardinality(df, column)
        print(column, "->", result.distinct_count)
        assert result.distinct_count == expected_count

    print("PASS: PySpark scalar dtypes")
else:
    print("SKIP: PySpark scalar dtypes")
```

## PySpark unsupported nested columns

```python
if spark is not None:
    schema = T.StructType([
        T.StructField("id", T.LongType(), False),
        T.StructField(
            "nested",
            T.ArrayType(T.LongType()),
            True,
        ),
    ])

    df = spark.createDataFrame(
        [
            (1, [1, 2]),
            (2, [3]),
            (3, [4, 5]),
        ],
        schema=schema,
    )

    report = profile_cardinality(df)

    outcomes = {
        outcome.column: outcome
        for outcome in report.columns
    }

    print("Rows:", report.total_row_count)
    print("Successful:", report.successful_column_count)
    print("Unsupported:", report.unsupported_column_count)

    assert outcomes["id"].status == "ok"
    assert outcomes["nested"].status == "unsupported"
    assert report.total_row_count == 3

    print("PASS: PySpark nested-column isolation")
else:
    print("SKIP: PySpark nested-column isolation")
```

## PySpark all-unsupported row-count edge case

```python
if spark is not None:
    schema = T.StructType([
        T.StructField(
            "array_col",
            T.ArrayType(T.IntegerType()),
            True,
        ),
        T.StructField(
            "map_col",
            T.MapType(
                T.StringType(),
                T.IntegerType(),
            ),
            True,
        ),
    ])

    df = spark.createDataFrame(
        [
            ([1], {"a": 1}),
            ([2], {"b": 2}),
            ([3], {"c": 3}),
        ],
        schema=schema,
    )

    report = profile_cardinality(df)

    print("Rows:", report.total_row_count)
    print("Successful:", report.successful_column_count)
    print("Unsupported:", report.unsupported_column_count)

    assert report.total_row_count == 3
    assert report.successful_column_count == 0
    assert report.unsupported_column_count == 2

    print("PASS: PySpark all-unsupported row count")
else:
    print("SKIP: PySpark all-unsupported row count")
```

# Cross-platform consistency

```python
values = [1, 2, 2, 3, 4, 4, 5]

pandas_df = pd.DataFrame({"id": values})
polars_df = pl.DataFrame({"id": values})

pandas_result = profile_cardinality(pandas_df, "id")
polars_result = profile_cardinality(polars_df, "id")

print("pandas:", pandas_result.distinct_count)
print("Polars:", polars_result.distinct_count)

assert pandas_result.distinct_count == 5
assert polars_result.distinct_count == 5

if spark is not None:
    spark_df = spark.createDataFrame(
        [(value,) for value in values],
        ["id"],
    )

    spark_result = profile_cardinality(spark_df, "id")
    print("PySpark:", spark_result.distinct_count)

    assert spark_result.distinct_count == 5
    print("PASS: pandas / Polars / PySpark consistency")
else:
    print("PySpark: skipped")
    print("PASS: pandas / Polars consistency")
```

# Cleanup

```python
if spark is not None:
    spark.stop()
    spark = None
    print("Spark stopped")
else:
    print("No Spark session to stop")
```
